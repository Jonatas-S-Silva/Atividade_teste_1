# **IMC**

Atividade para praticar

criar programa que calcula IMC

```python
# Etapa 1: peguntar o usuário seu dados.
print('#####  CALCULE SEU IMC  #####\n\n')
nome = input('Digite seu nome: ')
altura = input('\ndigite sua altura: ')
peso = input('\nDigite seu peso: ')
```

```python
# Etapa 2: converter variável.
peso_convert = float(peso)
altura_convert = float(altura.replace(',', '.'))
print(altura)
```

```python
# Etapa 3: Formula IMC.
imc = peso_convert / (altura_convert ** 2)
print(imc)
```

```python
# Etapa 4: responder o usuário.
if imc < 18.5:
  print(f'{nome}, você está abaixo do peso ideal.')
elif imc < 25:
  print(f'{nome}, você está no peso ideal.')
elif imc < 30:
  print(f'{nome}, você está com sobrepeso.')
else:
  print("Programa encerado!")
```

# **DESCONTO EM PRODUTO**

```python
print('#####  DESCONTO EM PRODUTO  #####\n\n')

nome_produto = input('Digite o nome do produto: ')
preco_original = float(input('Digite o preço original do produto: '))
desconto = float(input('Digite a porcentagem de desconto: '))

valor_desconto = preco_original * (desconto / 100)
valor_com_desconto = preco_original - valor_desconto

print('O produto: ', nome_produto,'\nque custava: R$',preco_original, '\ncom o desconto de ', desconto ,'%\npassou a custar: R$', valor_com_desconto)
```
